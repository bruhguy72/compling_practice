## 1.4. Строки: всё, что нужно для работы с текстом

Строка — это последовательность символов. У строк есть **методы** —
действия, которые можно с ними проделать. Метод пишется через точку после
строки.

````python
text = "  Лингвистика — это наука о языке.  "

print(f"Длина строки:        {len(text)}")
print(f"Убрать пробелы:      '{text.strip()}'")
print(f"Нижний регистр:      {text.lower()}")
print(f"Верхний регистр:     {text.upper()}")
print(f"Заменить:            {text.replace('наука', 'НАУКА')}")
print(f"Сколько раз 'а':     {text.count('а')}")
````

💬 Заметили? `len(text)` считает **все** символы, включая пробелы и
знаки препинания. Для компьютера пробел — такой же символ, как буква.

🔹 Важный метод: `split()` разбивает строку на части. По умолчанию — по
пробелам. Получается **список**.

````python
sentence = "Мама мыла раму"
words = sentence.split()

print(words)
print(f"Тип: {type(words)}")
print(f"Количество элементов: {len(words)}")
````

````python
print(" ".join(words))      # склеить через пробел
print("_".join(words))      # склеить через подчёркивание
print(" | ".join(words))    # склеить через вертикальную черту
````

## 1.7. Циклы

`for` повторяет действие для каждого элемента списка. Это главный
инструмент работы с текстом: почти всё, что мы делаем, — это «пройди по всем
словам и что-нибудь с ними сделай».

````python
text = "Лингвистика изучает язык как систему знаков"
words = text.split()

for word in words:
    print(f"{word:15} — {len(word)} букв")
````

💬 Запись `{word:15}` означает «выровняй по ширине 15 символов».
Мелочь, но таблица читается заметно лучше.

🔹 Цикл с условием внутри — уже полноценный фильтр.

````python
long_words = []

for word in words:
    if len(word) > 6:
        long_words.append(word)

print(long_words)
````

То же самое можно записать в одну строку — это называется **списковое
включение**. Читается почти как английская фраза: «слово для каждого слова
из words, если длина слова больше шести».

````python
long_words = [word for word in words if len(word) > 6]

print(long_words)
````

## 1.8. Функции

Функция — это кусок кода, которому дали имя, чтобы вызывать его много раз.
Мы уже пользовались готовыми (`print`, `len`), теперь напишем свою.

````python
def count_words(text):
    """Считает количество слов в тексте."""
    words = text.split()
    return len(words)


# проверяем
print(count_words("Мама мыла раму"))
print(count_words("Глокая куздра штеко будланула бокра"))
````

Разберём по частям:

- `def` — «определяю функцию»;
- `count_words` — имя, которое мы придумали;
- `(text)` — что функция принимает на вход;
- строка в тройных кавычках — описание функции для человека;
- `return` — что функция отдаёт обратно.

🔹 Функция может принимать несколько значений и иметь значения по умолчанию.

````python
def longest_word(text, min_length=1):
    """Возвращает самое длинное слово, не короче min_length."""
    words = [w for w in text.split() if len(w) >= min_length]

    if not words:
        return None

    return max(words, key=len)


print(longest_word("Лингвистика изучает язык"))
print(longest_word("кот и пёс"))
````

## 1.9. Словари и подсчёт частот

**Словарь** (`dict`) хранит пары «ключ — значение». Для нас это главная
структура: слово → сколько раз встретилось.

````python
frequencies = {"кот": 3, "собака": 2, "мышь": 1}

print(frequencies)
print(frequencies["кот"])           # получить значение по ключу
print(list(frequencies.keys()))     # все ключи
print(list(frequencies.values()))   # все значения
````

🔹 Посчитаем частоты вручную — чтобы увидеть механизм изнутри.

````python
text = "кот кот собака кот собака мышь"
words = text.split()

counter = {}

for word in words:
    if word in counter:
        counter[word] = counter[word] + 1
    else:
        counter[word] = 1

print(counter)
````

А теперь то же самое готовым инструментом. `Counter` из стандартной
библиотеки делает ровно это и умеет сразу отдавать самое частотное.

````python
from collections import Counter

counter = Counter(words)

print(counter)
print(counter.most_common(2))    # два самых частотных
````

💬 Так будет всю дорогу: сначала мы делаем что-то руками, чтобы понять
механизм, а потом берём готовый инструмент. Понимать механизм важно — иначе
непонятно, что делать, когда инструмент даёт странный результат.

## 1.10. Загружаем настоящий текст

Дальше нам нужен материал побольше. Есть три способа его получить.

**Способ 3. Скачать из интернета.** Ячейка ниже берёт статью из Википедии на русском языке. Если интернета нет, она молча подставит небольшой запасной текст,
и всё будет работать дальше.

````python
import json
import urllib.parse
import urllib.request

SAMPLE = """Язык устроен так, что небольшое число элементов порождает
бесконечное число сообщений. Из нескольких десятков звуков складываются
тысячи морфем, из морфем — сотни тысяч слов, из слов — предложения,
которых никто до нас не произносил. Эта способность языка называется
продуктивностью, и она отличает человеческую речь от всех известных
систем общения животных.

Лингвист смотрит на язык как на систему. Система эта многоуровневая:
фонетика описывает звуки, морфология — устройство слова, синтаксис —
устройство предложения, семантика — значение, прагматика — употребление.
Уровни связаны между собой, и изменение на одном отзывается на другом.

Компьютерная лингвистика добавляет к этому взгляду ещё одно измерение:
она спрашивает, какие из этих описаний можно превратить в алгоритм.
Оказалось, что далеко не все. Правила, которые человек применяет
не задумываясь, при попытке записать их формально рассыпаются на сотни
исключений. История обработки языка — это во многом история столкновения
красивых правил с живым материалом.

Первые системы машинного перевода строились на правилах. Лингвисты писали
словари соответствий и грамматики преобразований, и казалось, что задача
решается за несколько лет. Она не решилась. Выяснилось, что перевод
требует не только знания языка, но и знания о мире, а знание о мире
в правила не укладывается.

Следующий подход был статистическим. Вместо правил машине дали параллельные
тексты и позволили самой найти закономерности. Качество выросло, но
результат перестал быть объяснимым: система работала, а почему она выбрала
именно этот вариант перевода, сказать было трудно.

Современные модели пошли ещё дальше. Они не знают ни одного правила
в явном виде, зато видели такое количество текста, какое человек
не прочитает за всю жизнь. Они уверенно справляются с задачами, перед
которыми пасовали правила, и так же уверенно ошибаются там, где правило
сработало бы безотказно.

Для лингвиста это означает, что старые вопросы никуда не делись. Что такое
слово? Где проходит граница предложения? Одно и то же перед нами слово
или два разных? Ответы на эти вопросы приходится давать заново каждый раз,
когда мы берёмся обрабатывать новый материал. И отвечает на них не машина,
а исследователь."""


def load_wikipedia(title):
    """Скачивает текст статьи из Википедии на русском языке."""
    url = (
        "https://ru.wikipedia.org/w/api.php?action=query&prop=extracts"
        "&explaintext=1&format=json&titles=" + urllib.parse.quote(title)
    )

    with urllib.request.urlopen(url, timeout=10) as response:
        data = json.loads(response.read().decode("utf-8"))

    pages = data["query"]["pages"]
    page = next(iter(pages.values()))

    return page["extract"]


try:
    text = load_wikipedia("Лингвистика")
    print("Текст загружен из Википедии.")
except Exception as error:
    text = SAMPLE
    print("Не получилось скачать, работаем с запасным текстом.")
    print(f"(причина: {type(error).__name__})")

print(f"Длина текста: {len(text)} символов")
print()
print(text[:300], "...")
````

## 1.11. Собираем всё вместе: паспорт текста

🔹 Теперь у нас есть всё, чтобы написать функцию, упомянутую в начале занятия.

````python
def text_profile(text):
    """Возвращает основные характеристики текста."""
    words = text.lower().split()

    # уберём знаки препинания по краям слов — грубо, но пока сойдёт
    words = [w.strip(".,!?;:—«»()\"'") for w in words]
    words = [w for w in words if w]

    unique = set(words)
    avg_length = sum(len(w) for w in words) / len(words)

    return {
        "символов": len(text),
        "слов": len(words),
        "уникальных слов": len(unique),
        "средняя длина слова": round(avg_length, 2),
        "лексическое разнообразие": round(len(unique) / len(words), 3),
    }


profile = text_profile(text)

for key, value in profile.items():
    print(f"{key:28} {value}")
````

💬 Последний показатель — отношение числа уникальных слов к общему
числу слов. Его называют **TTR**, type-token ratio. Он показывает, насколько

# Регулярные выражения

Регулярное выражение — это способ описать **шаблон** строки. Не конкретное
слово, а правило, которому слово должно соответствовать: «начинается с
заглавной буквы», «состоит из четырёх цифр», «заканчивается на -ость».

Для лингвиста это очень естественный инструмент. По сути это формальная
грамматика для строк, и те, кто разбирался с формальными грамматиками,
схватывают её быстрее программистов.

Регулярные выражения — самый короткий путь к ощущению власти над текстом.
После этой пары вы сможете найти в тексте почти всё, что умеете описать
словами.

## 2.1. Первое совпадение

🔹 Всё, что нужно, — модуль `re` из стандартной библиотеки.

````python
import re

demo = "Кот сидел на окне. Другой кот спал. Третий кот ушёл."

print(re.findall("кот", demo))                       # найти все совпадения
print(re.findall("кот", demo, flags=re.IGNORECASE))  # без учёта регистра
````

💬 Первый вызов нашёл два «кот», а не три: «Кот» с заглавной буквы —
для компьютера другая строка. Флаг `re.IGNORECASE` это чинит.

Основные функции модуля:

| Функция | Что делает |
|---|---|
| `re.findall(шаблон, текст)` | список всех совпадений |
| `re.search(шаблон, текст)` | первое совпадение или `None` |
| `re.sub(шаблон, замена, текст)` | заменить все совпадения |
| `re.finditer(шаблон, текст)` | совпадения вместе с их позициями |

````python
match = re.search(r"кот", demo, flags=re.IGNORECASE)

print(match)
print(f"Найдено: '{match.group()}'")
print(f"Позиция: с {match.start()} по {match.end()}")
````

💬 Буква `r` перед кавычками означает «сырая строка»: в ней обратный
слеш остаётся обратным слешем. В регулярных выражениях слешей много, поэтому
`r"..."` пишут почти всегда. Привыкните сразу.

## 2.2. Классы символов

Вместо конкретной буквы можно указать **класс** — множество символов.

````python
sample = "В 2026 году на курсе 12 студентов, из них 3 отличника."

print(re.findall(r"\d", sample))        # одна цифра
print(re.findall(r"\d+", sample))       # одна или больше цифр подряд
print(re.findall(r"\w+", sample))       # последовательности «словесных» символов
print(re.findall(r"[аеёиоуыэюя]", sample.lower()))   # свой класс: гласные
````

| Класс | Что означает |
|---|---|
| `\d` | цифра |
| `\D` | не цифра |
| `\w` | буква, цифра или подчёркивание |
| `\W` | всё остальное |
| `\s` | пробельный символ |
| `\S` | не пробельный |
| `.` | любой символ, кроме перевода строки |
| `[абв]` | один из перечисленных |
| `[^абв]` | любой, кроме перечисленных |
| `[а-я]` | диапазон |

## 2.3. Квантификаторы: сколько раз

````python
words = "кот кооот коооооот кт кит"

print(re.findall(r"ко?т", words))        # 'о' ноль или один раз
print(re.findall(r"ко+т", words))        # 'о' один или больше раз
print(re.findall(r"ко*т", words))        # 'о' ноль или больше раз
print(re.findall(r"ко{2,4}т", words))    # 'о' от двух до четырёх раз
````

| Квантификатор | Сколько раз |
|---|---|
| `?` | ноль или один |
| `+` | один или больше |
| `*` | ноль или больше |
| `{3}` | ровно три |
| `{2,4}` | от двух до четырёх |
| `{2,}` | два и больше |

## 2.4. Якоря и границы

Якорь не совпадает с символом — он указывает на **место** в строке.

````python
words_list = ["пришёл", "прибыл", "приказ", "природа", "спринт"]

for word in words_list:
    if re.search(r"^при", word):
        print(f"{word:12} начинается на 'при'")
    else:
        print(f"{word:12} —")
````

| Якорь | Место |
|---|---|
| `^` | начало строки |
| `$` | конец строки |
| `\b` | граница слова |
| `\B` | не граница слова |

🔹 Граница слова `\b` — самый полезный якорь на практике.

````python
text2 = "Кот, кошка и котёнок сидели на коте... то есть на ковре."

print(re.findall(r"кот", text2, flags=re.IGNORECASE))       # ловит и внутри слов
print(re.findall(r"\bкот\b", text2, flags=re.IGNORECASE))   # только целое слово
````

## 2.5. Группы

Скобки выделяют часть шаблона. Тогда `findall` возвращает не всё совпадение,
а содержимое групп — это часто удивляет в начале.

````python
dates = "Занятия 12.09.2026, 19.09.2026 и 26.09.2026."

print(re.findall(r"\d{2}\.\d{2}\.\d{4}", dates))          # без групп
print(re.findall(r"(\d{2})\.(\d{2})\.(\d{4})", dates))    # с группами
````

## 2.6. Замена

`re.sub` заменяет найденное. В строке замены можно ссылаться на группы через
`\1`, `\2` и так далее.

````python
messy = "Здесь   слишком    много     пробелов ."

print(re.sub(r"\s+", " ", messy))              # схлопнуть пробелы
print(re.sub(r"\s+([.,!?])", r"\1", messy))    # убрать пробел перед знаком

# переставить дату из ДД.ММ.ГГГГ в ГГГГ-ММ-ДД
print(re.sub(r"(\d{2})\.(\d{2})\.(\d{4})", r"\3-\2-\1", dates))
````

## 3.2. Частотный список

````python
import re
from collections import Counter

try:
    text
except NameError:
    text = SAMPLE


def tokenize_simple(text):
    """Простая токенизация: только буквенные последовательности."""
    return re.findall(r"[а-яёa-z]+", text.lower())


words = tokenize_simple(text)
frequencies = Counter(words)

print(f"Всего слов:      {len(words)}")
print(f"Уникальных слов: {len(frequencies)}")
print()
print("Топ-20:")

for rank, (word, count) in enumerate(frequencies.most_common(20), start=1):
    share = count / len(words) * 100
    print(f"{rank:3}. {word:15} {count:5}   {share:5.2f}%")
````

````python
STOPWORDS = {
    "и", "в", "во", "не", "что", "он", "на", "я", "с", "со", "как", "а", "то",
    "все", "она", "так", "его", "но", "да", "ты", "к", "у", "же", "вы", "за",
    "бы", "по", "только", "ее", "мне", "было", "вот", "от", "меня", "еще",
    "нет", "о", "из", "ему", "теперь", "когда", "даже", "ну", "вдруг", "ли",
    "если", "уже", "или", "ни", "быть", "был", "него", "до", "вас", "нибудь",
    "уж", "вам", "ведь", "там", "потом", "себя", "ничего", "ей", "может",
    "они", "тут", "где", "есть", "надо", "ней", "для", "мы", "тебя", "их",
    "чем", "была", "сам", "чтоб", "без", "чего", "раз", "себе", "под",
    "будет", "тогда", "кто", "этот", "того", "потому", "этого", "какой",
    "совсем", "ним", "здесь", "этом", "один", "почти", "мой", "тем", "чтобы",
    "нее", "были", "куда", "зачем", "всех", "никогда", "можно", "при",
    "наконец", "два", "об", "другой", "хоть", "после", "над", "больше",
    "тот", "через", "эти", "нас", "про", "всего", "них", "какая", "много",
    "это", "её", "ещё",
}

content_words = [w for w in words if w not in STOPWORDS and len(w) > 2]
content_frequencies = Counter(content_words)

print("Топ-20 без служебных слов:")

for rank, (word, count) in enumerate(content_frequencies.most_common(20), start=1):
    print(f"{rank:3}. {word:20} {count}")
````

## 3.4. Закон Ципфа

Теперь самое интересное. Построим график: по горизонтали — ранг слова
(1 — самое частотное, 2 — второе по частоте и так далее), по вертикали —
его частота.

````python
import matplotlib.pyplot as plt

ranks = []
counts = []

for rank, (word, count) in enumerate(frequencies.most_common(), start=1):
    ranks.append(rank)
    counts.append(count)

plt.figure(figsize=(9, 5))
plt.plot(ranks, counts)
plt.xlabel("Ранг слова")
plt.ylabel("Частота")
plt.title("Распределение частот слов")
plt.grid(alpha=0.3)
plt.show()
````

💬 Получилась гипербола: несколько слов встречаются очень часто,
а огромный хвост — по одному разу. Такой график неудобно читать: почти всё
интересное сплющено у осей.

🔹 Тот же график в **логарифмических координатах**.

````python
plt.figure(figsize=(9, 5))
plt.loglog(ranks, counts, marker=".", linestyle="none", markersize=4)
plt.xlabel("Ранг слова (лог. шкала)")
plt.ylabel("Частота (лог. шкала)")
plt.title("Закон Ципфа")
plt.grid(alpha=0.3, which="both")
plt.show()
````

**Получилась прямая.**

Это и есть закон Ципфа: частота слова обратно пропорциональна его рангу.
Второе по частоте слово встречается примерно вдвое реже первого, третье —
втрое реже, десятое — в десять раз реже.

Формально: *частота × ранг ≈ константа*, или *f(r) ≈ C / r^a*, где показатель
`a` близок к единице.

Самое поразительное в этом законе — его универсальность. Он выполняется:

- для всех естественных языков, включая те, что не родственны друг другу;
- для текстов любого жанра и любой длины;
- для языков, которые ещё не расшифрованы, — это один из аргументов
  в пользу того, что перед нами язык, а не орнамент;
- за пределами языка: для размеров городов, доходов, посещаемости сайтов.

Общепринятого объяснения, почему так, до сих пор нет. Есть несколько
конкурирующих гипотез, и это открытый вопрос.

🔹 Проверим численно: посчитаем произведение ранга на частоту для первых
двадцати слов. Если закон работает, числа должны быть одного порядка.

````python
print("ранг × частота для первых 20 слов:")
print()

for rank, (word, count) in enumerate(frequencies.most_common(20), start=1):
    print(f"{rank:3}. {word:15} {count:5} × {rank:3} = {count * rank:6}")
````

### ⭐ 2. Закон Хипса

Второй эмпирический закон: число уникальных слов растёт как корень из числа
всех слов. То есть словарь растёт медленнее текста, и чем дальше, тем
медленнее.

Постройте график: по горизонтали — сколько слов текста прочитано, по
вертикали — сколько уникальных слов встретилось к этому моменту.

````python
seen = set()
vocabulary_growth = []

for word in words:
    seen.add(word)
    vocabulary_growth.append(len(seen))

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(words) + 1), vocabulary_growth)
plt.xlabel("Прочитано слов")
plt.ylabel("Уникальных слов")
plt.title("Закон Хипса: рост словаря")
plt.grid(alpha=0.3)
plt.show()

print(f"На {len(words)} словах словарь составил {len(seen)} единиц")
````

💬 Отсюда прямое следствие для практики: **TTR нельзя использовать для
сравнения текстов разной длины**. Чем длиннее текст, тем ниже TTR — не
потому, что автор беднее, а потому, что так устроен язык.

Существуют меры, устойчивые к длине: MTLD, MATTR, мера Ги́ро. Если ваш
итоговый проект будет про лексическое разнообразие, брать нужно их.
